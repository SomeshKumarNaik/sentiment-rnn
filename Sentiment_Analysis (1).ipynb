## Pre-Processing

### i) Converting to lowercase

```python
df["review"] = df["review"].str.lower()
```

```python
df["review"]
```

### ii) Removing URLs

```python
import re
```

```python
def remove_urls(text):
    text = re.sub(r"http\S+", "", text)
    return text

df["review"] = df["review"].apply(remove_urls)
```

### iii) Removing punctuations

```python
def remove_punctuations(text):
    text = re.sub(r"[^A-Za-z0-9\s]", "", text)
    return text

df["review"] = df["review"].apply(remove_punctuations)
```

```python
df
```

### iv) Removing HTML

```python
def remove_html(text):
    text = re.sub(r"<.*?>", "", text)
    return text

df["review"] = df["review"].apply(remove_html)
```

```python
df["review"]
```

### v) Removing Stopwords

```python
import nltk
```

```python
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")
```

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
```

```python
def remove_stopwords(text):
    tokens = word_tokenize(text)
    stop_words = stopwords.words("english")

    for word in tokens:
        if word in stop_words:
            text = text.replace(word, "")
    return text
```

```python
df["review"] = df["review"].apply(remove_stopwords)
```

```python
df["review"]
```

### vi) Stemming

```python
from nltk.stem import PorterStemmer
```

```python
def stemming(text):
    ps = PorterStemmer()
    stemmed_words = []

    tokens = word_tokenize(text)
    for token in tokens:
        stemmed_token = ps.stem(token)
        stemmed_words.append(stemmed_token)

    return " ".join(stemmed_words)
```

```python
df["review"] = df["review"].apply(stemming)
```

```python
df["review"]
```

### vii) Encoding

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df["sentiment"] = le.fit_transform(df["sentiment"])

y = df["sentiment"] 
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tf = TfidfVectorizer(max_features = 5000)

x = tf.fit_transform(df["review"])
```

## Dataset and DataLoaders

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
import torch
from torch.utils.data import TensorDataset, DataLoader
```

```python
x_train = x_train.toarray()
x_test = x_test.toarray()
```

```python
train_set = TensorDataset(
    torch.from_numpy(x_train).float(),
    torch.from_numpy(y_train.values).float()
)

test_set = TensorDataset(
    torch.from_numpy(x_test).float(),
    torch.from_numpy(y_test.values).float()
)
```

```python
train_loader = DataLoader(train_set, shuffle = True, batch_size = 64)
test_loader = DataLoader(test_set, shuffle = True, batch_size = 64)
```

## Building the RNN

```python
import torch.nn as nn
import torch.optim as optimizer
```

```python
class RNN(nn.Module):
    def __init__(self, input_size, hidden_size = 128, num_layers = 1):
        super().__init__()

        self.hidden_size = hidden_size
        self.num_layers = num_layers

        # RNN layer
        self.rnn = nn.RNN(input_size, hidden_size, num_layers, batch_first = True)

        # Fully Connected layer
        self.fc = nn.Linear(hidden_size, 1)

    def forward(self, x):
        h0 = torch.zeros(self.num_layers, x.size(0), self.hidden_size)
        out, _ = self.rnn(x, h0)
        out = self.fc(out[:, -1, :])
        
        return out
```

## Model

```python
input_size = x_train.shape[1]

model = RNN(input_size)

critetion = nn.BCELoss()

optimizer = optimizer.Adam(model.parameters())
```

## Training the RNN

```python
epochs = 10

for epoch in range(epochs):
    model.train()

    for xb, yb in train_loader:
        optimizer.zero_grad

        xb = xb.unsqueeze(1)
        outputs = model(xb)

        outputs = torch.sigmoid(outputs.squeeze())

        loss = critetion(outputs, yb)

        loss.backward()

        optimizer.step()
        print(f"{epoch + 1}/{epochs} and loss = {loss.item()}")
```

## Evaluation

```python
model.eval()

with torch.no_grad():
    correct_vals = 0
    total_vals = 0

    for xb, yb in test_loader:
        xb = xb.unsqueeze(1)
        outputs = model(xb)
        predicted = (torch.sigmoid(outputs.squeeze()) > 0.5).float()

        total_vals += yb.size(0)
        correct_vals += (predicted == yb).sum().item()

    print(f"accuracy = {(correct_vals / total_vals)*100}")
```
